# Run Verb-Object-Extraction with Sample Dataset

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    handlers=[
        logging.StreamHandler()
    ]
)

import pandas as pd
from verb_object_extraction import search_pairs_in_jobads
```

## Read jobad sample

```python
jobad_df = pd.read_csv('../data/sample_jobads.csv')
print(len(jobad_df))
jobad_df.head()
```

## Search verb-object-pairs within plain text

```python
pairs_df = search_pairs_in_jobads(jobad_df, text_column='jobad_text')
print(len(pairs_df))
pairs_df.head()
```

## How many verb-object-pairs are in which task dimension?

```python
pairs_df['Task_Dimension'].value_counts()
```

## Which verb is assigned to which task dimension?

```python
pd.crosstab(pairs_df['verb'], pairs_df['Task_Dimension']).sort_values('Interactive', ascending=False)
```
